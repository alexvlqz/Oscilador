# Oscilador Armónico Simple

Construir un programa que simule un Oscilador Armónico Simple.


---

Por convención definimos $\ddot{x}\equiv a$ y $\dot{x}\equiv v$.

Teniendo un sistema masa-resorte, la fuerza ejercida se define como
$$
F_R=-kx
$$
Aplicando la segunda ley de Newton,
$$
\Sigma F=m\ddot{x}\implies-kx=m\ddot{x}\implies\ddot{x}+\dfrac{k}{m} x=0
$$
Y definiendo $\omega_0^2\equiv k/m$, donde a $\omega_0$ es la frecuencia angular del sistema, obtenemos la ecuación de movimiento
$$
\ddot{x}+\omega_0^2 x=0
$$
Al resolver la ecuación diferencial, tenemos por resultado
$$
x(t)=A_0cos(\omega_0t+\phi)
$$
Evaluamos la función usando Python:

```python
# Importamos las dependencias
import numpy as np
import matplotlib.pyplot as plt
```

Evaluamos la función para valores dados $A_0=1$, $w_0=1$, y $\phi=0$, evaluando en $D(x)=[0,2\pi]$

```python
t = np.arange(0.0,2*np.pi,0.01)
x = np.cos(t)
```

Teniendo los valores evaluados, creamos una gráfica usando MatPlotLib

```python
fig, ax = plt.subplots()
ax.plot(t, x)

ax.set(xlabel='t', ylabel='x(t)',
       title='Oscilador Harmónico Simple')

plt.show()
```

```python
import matplotlib.animation as animation
from IPython.display import HTML
```

```python
fig, ax = plt.subplots()


t = np.arange(0, 2*np.pi, 0.01)
line, = ax.plot(t, np.sin(t))
ax.set(xlabel='t', ylabel='x(t)',
       title='Oscilador Harmónico Simple')

def animate(i):
    line.set_ydata(np.sin(t + i / 50))  # update the data.
    return line,


ani = animation.FuncAnimation(
    fig, animate, interval=20, blit=True, save_count=500)

# To save the animation, use e.g.
#
# ani.save("movie.mp4")
#
# or
#
# writer = animation.FFMpegWriter(
#     fps=15, metadata=dict(artist='Me'), bitrate=1800)
# ani.save("movie.mp4", writer=writer)

HTML(ani.to_jshtml())
```
